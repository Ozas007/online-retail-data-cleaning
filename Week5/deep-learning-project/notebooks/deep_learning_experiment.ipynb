# Week 5 — Deep Learning Experiment: MLP on MNIST

This notebook complements the modular `src.main` pipeline. It loads the built-in MNIST dataset via `tf.keras.datasets.mnist.load_data()`, visualises a handful of raw digits, and documents the overall architecture and evaluation flow before the main script trains end-to-end.

## 1. Imports and configuration

Run the pipeline end-to-end with `python -m src.main`. This notebook only demonstrates the dataset and helper-level utilities.

```python
# Core imports (no PyTorch — TensorFlow/Keras only)
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

import tensorflow as tf
from tensorflow.keras.datasets import mnist

PROJECT_ROOT = Path.cwd().resolve()
sys.path.insert(0, str(PROJECT_ROOT))

from src.config import (
    RANDOM_SEED, NUM_CLASSES, INPUT_SHAPE, EPOCHS, BATCH_SIZE, LEARNING_RATE,
    VALIDATION_SPLIT, EARLY_STOPPING_PATIENCE, DROPOUT_RATE,
)

np.random.seed(RANDOM_SEED)
tf.random.set_seed(RANDOM_SEED)

print(f"PROJECT_ROOT       = {PROJECT_ROOT}")
print(f"TF version         = {tf.__version__}")
print(f"RANDOM_SEED        = {RANDOM_SEED}")
print(f"INPUT_SHAPE        = {INPUT_SHAPE}")
print(f"NUM_CLASSES        = {NUM_CLASSES}")
print(f"EPOCHS / BATCH     = {EPOCHS} / {BATCH_SIZE}")
print(f"LR / VAL / PATIENCE= {LEARNING_RATE} / {VALIDATION_SPLIT} / {EARLY_STOPPING_PATIENCE}")
print(f"DROPOUT_RATE       = {DROPOUT_RATE}")
```

```python
# 2. Load MNIST from tf.keras built-in datasets
(X_train, y_train), (X_test, y_test) = mnist.load_data()

print(f"X_train = {X_train.shape}  dtype={X_train.dtype}  range=[{X_train.min()}, {X_train.max()}]")
print(f"y_train = {y_train.shape}  dtype={y_train.dtype}  unique={np.unique(y_train)}")
print(f"X_test  = {X_test.shape}   dtype={X_test.dtype}")
print(f"y_test  = {y_test.shape}   dtype={y_test.dtype}")

n_show = 8
fig, axes = plt.subplots(1, n_show, figsize=(2 * n_show, 2))
for i, ax in enumerate(axes):
    ax.imshow(X_train[i], cmap="gray")
    ax.set_title(f"y = {int(y_train[i])}")
    ax.axis("off")
fig.suptitle("Raw MNIST samples (first 8 training images)", fontsize=12)
plt.tight_layout()
plt.show()
```
